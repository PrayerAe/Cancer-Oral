```python
import os
import shutil

# --- PENGATURAN ---
# 1. Ganti path ini sesuai dengan lokasi folder Anda
folder_sumber_1 = "../../dataset/Oral Cancer Dataset/CANCER"
folder_sumber_2 = "../../dataset/OC Dataset kaggle new/CANCER_V2"
folder_tujuan = "../../dataset/OralCancer_V2"
# --------------------

# Daftar semua folder sumber yang akan diproses
list_folder_sumber = [folder_sumber_1, folder_sumber_2]

# Membuat folder tujuan jika belum ada
if not os.path.exists(folder_tujuan):
    os.makedirs(folder_tujuan)
    print(f"Folder tujuan dibuat: {folder_tujuan}")

# Loop melalui setiap folder sumber
for folder_path in list_folder_sumber:
    nama_folder_sumber = os.path.basename(folder_path)

    # Loop melalui setiap file di dalam folder sumber
    for nama_file in os.listdir(folder_path):
        nama_file_baru = f"{nama_folder_sumber}_{nama_file}"

        # Membuat path lengkap untuk file sumber dan tujuan
        path_sumber = os.path.join(folder_path, nama_file)
        path_tujuan = os.path.join(folder_tujuan, nama_file_baru)

        # Menyalin file dari sumber ke tujuan dengan nama baru
        shutil.copy2(path_sumber, path_tujuan)
        print(f"Disalin: {nama_file} -> {nama_file_baru}")

print("\n✅ Proses penggabungan selesai!")
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# --- PENGATURAN PATH (samakan dengan sebelumnya) ---
folder_sumber_1 = r"../../dataset/Oral Cancer Dataset/NON-CANCER"
folder_sumber_2 = r"../../dataset/OC Dataset kaggle new/NON CANCER_V2"
folder_tujuan   = r"../../dataset/OralCancer_DatasetGabungan/NON-CANCER"

# Daftar folder sumber
list_folder_sumber = [folder_sumber_1, folder_sumber_2]

# --- Hitung jumlah file di masing-masing folder ---
jumlah_data = {}
for folder in list_folder_sumber:
    nama = os.path.basename(folder)
    jumlah_data[nama] = len(os.listdir(folder))

# Hitung jumlah total setelah digabung
jumlah_data["Setelah Digabung"] = len(os.listdir(folder_tujuan))

# --- Visualisasi ---
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.barplot(x=list(jumlah_data.keys()), y=list(jumlah_data.values()), palette="viridis")

plt.title("Perbandingan Jumlah Gambar Sebelum dan Sesudah Digabung", fontsize=14)
plt.xlabel("Folder", fontsize=12)
plt.ylabel("Jumlah Gambar", fontsize=12)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Path ke folder gabungan (final dataset)
folder_cancer = r"../../dataset/OralCancer_DatasetGabungan/CANCER"
folder_non_cancer = r"../../dataset/OralCancer_DatasetGabungan/NON-CANCER"

# Hitung jumlah file di masing-masing kelas
jumlah_per_kelas = {
    "CANCER": len(os.listdir(folder_cancer)),
    "NON-CANCER": len(os.listdir(folder_non_cancer))
}

# --- Visualisasi ---
sns.set(style="whitegrid")
plt.figure(figsize=(8, 6))
barplot = sns.barplot(x=list(jumlah_per_kelas.keys()),
                      y=list(jumlah_per_kelas.values()),
                      palette="mako")

# Tambahkan angka di atas setiap bar
for i, value in enumerate(jumlah_per_kelas.values()):
    plt.text(i, value + 5, str(value), ha='center', va='bottom', fontsize=12)

plt.title("Distribusi Data Setelah Penggabungan", fontsize=14)
plt.xlabel("Kelas", fontsize=12)
plt.ylabel("Jumlah Gambar", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Total data setelah oversampling
total_data = 1980

# Hitung pembagian data
jumlah_split = {
    "Training (80%)": int(0.80 * total_data),
    "Validation (15%)": int(0.15 * total_data),
    "Testing (5%)": total_data - int(0.80 * total_data) - int(0.15 * total_data)
}

# --- Visualisasi ---
sns.set(style="whitegrid")
plt.figure(figsize=(9, 6))
barplot = sns.barplot(x=list(jumlah_split.keys()),
                      y=list(jumlah_split.values()),
                      palette="flare")

# Tambahkan label jumlah di atas setiap bar
for i, value in enumerate(jumlah_split.values()):
    plt.text(i, value + 10, str(value), ha='center', va='bottom', fontsize=12)

plt.title("Pembagian Dataset Setelah Oversampling (Total: 1980 Gambar)", fontsize=14)
plt.xlabel("Subset Dataset", fontsize=12)
plt.ylabel("Jumlah Gambar", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os
import random
import time
import warnings
from datetime import datetime

import torch
from torch.utils.data import DataLoader, Dataset
import torch.nn as nn
from torchvision import transforms, models
from sklearn.model_selection import train_test_split
from PIL import Image

warnings.filterwarnings("ignore")
```

```python
def seed_everything(seed):
    """Mengatur seed untuk reproduktifitas."""
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.backends.cudnn.deterministic = True

seed_everything(42)
```

```python
hyperparameters = {
    'num_classes': 2,
    'learning_rate': 1e-4,
    'img_size': 224, 
    'batch_size': 32,
    'n_epochs': 30,
    'data_path': r'../../dataset/OralCancer_DatasetGabungan',
}
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
def load_data_from_folders(data_path):
    """Memuat path gambar dan label dari struktur folder."""
    image_paths = []
    labels = []
    class_map = {'Cancer': 1, 'Non-Cancer': 0}
    
    for class_name, label in class_map.items():
        class_dir = os.path.join(data_path, class_name)
        if not os.path.isdir(class_dir):
            print(f"Warning: Folder '{class_dir}' tidak ditemukan.")
            continue
        for img_name in os.listdir(class_dir):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                image_paths.append(os.path.join(class_dir, img_name))
                labels.append(label)
                
    return image_paths, labels

all_image_paths, all_labels = load_data_from_folders(hyperparameters['data_path'])

if not all_image_paths:
    raise ValueError("Tidak ada gambar yang ditemukan. Periksa 'data_path' dan struktur folder Anda.")
```

```python
import collections

# Cek distribusi kelas pada KESELURUHAN DATA SEBELUM oversampling
print("Distribusi KESELURUHAN data SEBELUM oversampling:")
# Variabel yang digunakan adalah 'all_labels'
label_counts_before = collections.Counter(all_labels)
print(f"  - Non-Cancer (0): {label_counts_before[0]} sampel")
print(f"  - Cancer (1): {label_counts_before[1]} sampel")

# Tentukan kelas minoritas dan mayoritas
minority_class_label = 0  # Non-Cancer
majority_class_label = 1  # Cancer

count_majority = label_counts_before[majority_class_label]
count_minority = label_counts_before[minority_class_label]

# Lakukan oversampling jika diperlukan
if count_minority < count_majority:
    num_to_add = count_majority - count_minority
    print(f"\nKelas 'Non-Cancer' adalah minoritas. Menambahkan {num_to_add} sampel ke seluruh dataset...")

    # Dapatkan path dari semua sampel kelas minoritas di seluruh dataset
    # Variabel yang digunakan adalah 'all_image_paths' dan 'all_labels'
    minority_paths = [path for path, label in zip(all_image_paths, all_labels) if label == minority_class_label]
    
    # Duplikasi sampel minoritas
    oversampled_paths_to_add = random.choices(minority_paths, k=num_to_add)
    oversampled_labels_to_add = [minority_class_label] * num_to_add

    # Tambahkan sampel duplikat ke list data utama
    # Variabel yang diubah adalah 'all_image_paths' dan 'all_labels'
    all_image_paths.extend(oversampled_paths_to_add)
    all_labels.extend(oversampled_labels_to_add)

    # Kocok kembali keseluruhan data agar terdistribusi acak
    temp_combined = list(zip(all_image_paths, all_labels))
    random.shuffle(temp_combined)
    all_image_paths, all_labels = zip(*temp_combined)
    
    # Konversi kembali ke list
    all_image_paths = list(all_image_paths)
    all_labels = list(all_labels)

    # Cek distribusi kelas SETELAH oversampling
    print("\nDistribusi KESELURUHAN data SETELAH oversampling:")
    label_counts_after = collections.Counter(all_labels)
    print(f"  - Non-Cancer (0): {label_counts_after[0]} sampel")
    print(f"  - Cancer (1): {label_counts_after[1]} sampel")
    print(f"Total data baru: {len(all_image_paths)} sampel\n")
```

```python
train_paths, temp_paths, train_labels, temp_labels = train_test_split(
    all_image_paths, all_labels, test_size=0.2, random_state=42, stratify=all_labels
)
val_paths, test_paths, val_labels, test_labels = train_test_split(
    temp_paths, temp_labels, test_size=0.25, random_state=42, stratify=temp_labels
)

print(f"Total gambar: {len(all_image_paths)}")
print(f"Data Latih: {len(train_paths)} sampel")
print(f"Data Validasi: {len(val_paths)} sampel")
print(f"Data Uji: {len(test_paths)} sampel")
```

```python
class OralCancerDataset(Dataset):
    def __init__(self, image_paths, labels, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        label = torch.tensor(self.labels[idx], dtype=torch.long)
        
        try:
            image = Image.open(img_path).convert("RGB")
            if self.transform:
                image = self.transform(image)
        except Exception as e:
            print(f"Gagal memuat atau mentransformasi gambar: {img_path}. Error: {e}")
            # Mengembalikan tensor kosong jika terjadi kesalahan
            return torch.zeros((3, hyperparameters['img_size'], hyperparameters['img_size'])), label

        return image, label
```

```python
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(hyperparameters['img_size'], scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(30), # Tingkatkan rotasi dari 20 menjadi 30 derajat
    transforms.RandomAffine(degrees=0, translate=(0.1, 0.1), scale=(0.9, 1.1)),
    transforms.ColorJitter(brightness=0.4, contrast=0.4, saturation=0.4, hue=0.1), # Tingkatkan intensitas
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

val_test_transform = transforms.Compose([
    transforms.Resize((hyperparameters['img_size'], hyperparameters['img_size'])),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])
```

```python
train_dataset = OralCancerDataset(train_paths, train_labels, transform=train_transform)
val_dataset = OralCancerDataset(val_paths, val_labels, transform=val_test_transform)
test_dataset = OralCancerDataset(test_paths, test_labels, transform=val_test_transform)

train_loader = DataLoader(train_dataset, batch_size=hyperparameters['batch_size'], shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=hyperparameters['batch_size'], shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=hyperparameters['batch_size'], shuffle=False)

print(f'Ukuran Train loader: {len(train_loader)}')
print(f'Ukuran Validation loader: {len(val_loader)}')
print(f'Ukuran Test loader: {len(test_loader)}')
```

```python
# In[11]: Versi VGG16 dengan classifier yang lebih sederhana dan regularisasi lebih kuat

# Muat model VGG16
model = models.vgg16(pretrained=True)

# Bekukan lapisan fitur
for param in model.features.parameters():
    param.requires_grad = False

# Dapatkan jumlah fitur input
num_ftrs = model.classifier[0].in_features # -> 25088

# Ganti classifier dengan yang JAUH LEBIH SEDERHANA dan dropout lebih tinggi
model.classifier = nn.Sequential(
    nn.Linear(num_ftrs, 512),  # Kurangi jumlah neuron dari 4096 menjadi 512
    nn.ReLU(inplace=True),
    nn.Dropout(0.5), # Tambahkan dropout yang lebih tinggi (0.5 adalah nilai yang baik)
    nn.Linear(512, hyperparameters['num_classes']) # Langsung ke output
)

# Pindahkan model ke device
model = model.to(device)

print("Arsitektur classifier VGG16 baru yang lebih sederhana:")
print(model.classifier)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    filter(lambda p: p.requires_grad, model.parameters()), 
    lr=hyperparameters['learning_rate'],
    weight_decay=5e-4
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=hyperparameters['n_epochs'],
    eta_min=1e-6
)
```

```python
def train_and_validate_model(model, train_loader, val_loader, criterion, optimizer, scheduler, hyperparameters, patience=7):
    # Inisialisasi
    total_start_time = time.time()
    best_val_loss = float('inf')
    best_val_acc = 0.0
    patience_counter = 0

    # Setup direktori simpan model
    model_name = hyperparameters.get('model_name', 'vgg16s')  
    base_folder = "../../models"
    checkpoint_folder = os.path.join(base_folder, "checkpoints")
    os.makedirs(checkpoint_folder, exist_ok=True)

    # Device
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)

    # Riwayat training
    history = {
        'train_loss': [], 'val_loss': [],
        'train_acc': [], 'val_acc': []
    }

    # Training loop
    for epoch in range(hyperparameters['n_epochs']):
        epoch_start_time = time.time()
        
        # ------------------- TRAINING -------------------
        model.train()
        total_train_loss, total_train_correct = 0.0, 0

        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            total_train_loss += loss.item() * inputs.size(0)
            total_train_correct += (outputs.argmax(1) == labels).sum().item()

        avg_train_loss = total_train_loss / len(train_loader.dataset)
        train_accuracy = total_train_correct / len(train_loader.dataset) * 100
        history['train_loss'].append(avg_train_loss)
        history['train_acc'].append(train_accuracy)

        # ------------------- VALIDATION -------------------
        model.eval()
        total_val_loss, total_val_correct = 0.0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                total_val_loss += loss.item() * inputs.size(0)
                total_val_correct += (outputs.argmax(1) == labels).sum().item()

        avg_val_loss = total_val_loss / len(val_loader.dataset)
        val_accuracy = total_val_correct / len(val_loader.dataset) * 100
        history['val_loss'].append(avg_val_loss)
        history['val_acc'].append(val_accuracy)

        scheduler.step()

        # ------------------- LOG -------------------
        epoch_time = time.time() - epoch_start_time
        print(f"Epoch {epoch + 1}/{hyperparameters['n_epochs']} | "
              f"Train Loss: {avg_train_loss:.4f}, Acc: {train_accuracy:.2f}% | "
              f"Val Loss: {avg_val_loss:.4f}, Acc: {val_accuracy:.2f}% | "
              f"Time: {epoch_time:.2f}s")

        # ------------------- CHECKPOINT -------------------
        if val_accuracy > best_val_acc:
            best_val_acc = val_accuracy
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            filename = f"{model_name}_epoch{epoch+1}_valacc{val_accuracy:.4f}_{timestamp}.pth"
            save_path = os.path.join(checkpoint_folder, filename)
            torch.save(model.state_dict(), save_path)
            print(f"✅ Model disimpan: {save_path}")
            patience_counter = 0  # reset patience
        else:
            patience_counter += 1
            print(f"No improvement. Patience: {patience_counter}/{patience}")

        # ------------------- EARLY STOPPING -------------------
        if patience_counter >= patience:
            print("⛔ Early stopping triggered.")
            break

    # ------------------- SELESAI -------------------
    total_training_time = time.time() - total_start_time
    print(f"\n🕒 Total Training Time: {total_training_time / 60:.2f} minutes")
    return history
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
```

```python
def evaluate_model(model, test_loader):
    """Evaluasi model pada test set."""
    model.eval()
    all_preds = []
    all_labels = []
    test_correct = 0
    
    with torch.no_grad():
        for inputs, labels in test_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
            test_correct += (preds == labels).sum().item()
    
    test_accuracy = test_correct / len(test_loader.dataset) * 100
    print(f"Test Accuracy: {test_accuracy:.2f}%")
    
    return all_preds, all_labels, test_accuracy
```

```python
def plot_confusion_matrix(y_true, y_pred, class_names):
    """Plot confusion matrix dengan matplotlib."""
    cm = confusion_matrix(y_true, y_pred)
    
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # Membuat heatmap manual
    im = ax.imshow(cm, interpolation='nearest', cmap='Blues')
    ax.figure.colorbar(im, ax=ax)
    
    # Set ticks
    ax.set(xticks=np.arange(cm.shape[1]),
           yticks=np.arange(cm.shape[0]),
           xticklabels=class_names,
           yticklabels=class_names,
           title='Confusion Matrix',
           ylabel='True Label',
           xlabel='Predicted Label')
    
    # Rotate tick labels
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
    
    # Menambahkan text annotations
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis] * 100
    thresh = cm.max() / 2.
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, f'{cm[i, j]}\n({cm_normalized[i, j]:.1f}%)',
                   ha="center", va="center",
                   color="white" if cm[i, j] > thresh else "black",
                   fontsize=12)
    
    plt.tight_layout()
    plt.savefig('VGG16s_confusion_matrix.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    return cm
```

```python
def calculate_metrics(y_true, y_pred, class_names):
    """Menghitung dan menampilkan berbagai metrik evaluasi."""
    cm = confusion_matrix(y_true, y_pred)
    
    # Menghitung metrik per kelas
    tn, fp, fn, tp = cm.ravel()
    
    # Metrik untuk kelas positif (Cancer)
    sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0  # Recall/Sensitivity
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    f1_score = 2 * (precision * sensitivity) / (precision + sensitivity) if (precision + sensitivity) > 0 else 0
    accuracy = (tp + tn) / (tp + tn + fp + fn)
    
    print("\n" + "="*50)
    print("HASIL EVALUASI MODEL")
    print("="*50)
    
    print(f"\nOverall Accuracy: {accuracy*100:.2f}%")
    print(f"\nUntuk Kelas 'Cancer' (Positive Class):")
    print(f"  - Sensitivity (Recall): {sensitivity*100:.2f}%")
    print(f"  - Specificity: {specificity*100:.2f}%")
    print(f"  - Precision: {precision*100:.2f}%")
    print(f"  - F1-Score: {f1_score*100:.2f}%")
    
    print(f"\nConfusion Matrix Details:")
    print(f"  - True Positives (Cancer detected correctly): {tp}")
    print(f"  - True Negatives (Non-Cancer detected correctly): {tn}")
    print(f"  - False Positives (Non-Cancer misclassified as Cancer): {fp}")
    print(f"  - False Negatives (Cancer misclassified as Non-Cancer): {fn}")
    
    # Classification report
    print("\n" + "="*50)
    print("CLASSIFICATION REPORT")
    print("="*50)
    print(classification_report(y_true, y_pred, target_names=class_names))
    
    return {
        'accuracy': accuracy,
        'sensitivity': sensitivity,
        'specificity': specificity,
        'precision': precision,
        'f1_score': f1_score,
        'confusion_matrix': cm
    }
```

```python
def visualize_predictions(model, test_dataset, num_samples=8):
    """Visualisasi beberapa sampel prediksi."""
    model.eval()
    class_names = ['Non-Cancer', 'Cancer']
    
    # Ambil sampel random
    indices = random.sample(range(len(test_dataset)), min(num_samples, len(test_dataset)))
    
    fig, axes = plt.subplots(2, 4, figsize=(15, 8))
    axes = axes.ravel()
    
    with torch.no_grad():
        for idx, i in enumerate(indices):
            if idx >= len(axes):
                break
                
            image, true_label = test_dataset[i]
            image_display = image.clone()
            
            # Denormalisasi untuk display
            mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
            std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)
            image_display = image_display * std + mean
            image_display = torch.clamp(image_display, 0, 1)
            
            # Prediksi
            image = image.unsqueeze(0).to(device)
            output = model(image)
            probabilities = torch.nn.functional.softmax(output, dim=1)
            _, pred = torch.max(output, 1)
            
            # Plot
            axes[idx].imshow(image_display.permute(1, 2, 0))
            axes[idx].axis('off')
            
            true_class = class_names[true_label]
            pred_class = class_names[pred.item()]
            prob = probabilities[0][pred.item()].item() * 100
            
            color = 'green' if true_label == pred.item() else 'red'
            axes[idx].set_title(f'True: {true_class}\nPred: {pred_class} ({prob:.1f}%)', 
                               color=color, fontsize=10)
    
    plt.tight_layout()
    plt.savefig('VGG16s_sample_predictions.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def overlay_heatmap_on_image(image_path, mask_path, alpha=0.5, colormap=cv2.COLORMAP_JET):

    # Load gambar asli dan mask
    image = cv2.imread(image_path)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

    # Resize agar ukurannya sama
    mask = cv2.resize(mask, (image.shape[1], image.shape[0]))

    # Normalize mask ke [0, 255]
    normalized_mask = cv2.normalize(mask, None, 0, 255, cv2.NORM_MINMAX)

    # Convert colormap ke warna
    heatmap = cv2.applyColorMap(normalized_mask.astype(np.uint8), colormap)

    # Overlay heatmap ke gambar asli
    overlay = cv2.addWeighted(image, 1 - alpha, heatmap, alpha, 0)

    # Convert BGR ke RGB untuk ditampilkan dengan matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    overlay_rgb = cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)

    # Plot hasilnya
    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(image_rgb)
    plt.title('Gambar Asli')
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(mask, cmap='gray')
    plt.title('Mask / Segmentasi')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(overlay_rgb)
    plt.title('Overlay Heatmap')
    plt.axis('off')

    plt.tight_layout()
    plt.savefig('vg16s_heatmap_overlay_comparison.png', dpi=300)
    plt.show()

    return image_rgb, mask, overlay_rgb
```

```python
class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer

        self.gradients = None
        self.activations = None

        # Hook untuk simpan gradient dan activation
        target_layer.register_forward_hook(self.forward_hook)
        target_layer.register_backward_hook(self.backward_hook)

    def forward_hook(self, module, input, output):
        self.activations = output.detach()

    def backward_hook(self, module, grad_input, grad_output):
        self.gradients = grad_output[0].detach()

    def generate(self, input_tensor, class_idx=None):
        self.model.eval()
        output = self.model(input_tensor)

        if class_idx is None:
            class_idx = torch.argmax(output, dim=1).item()

        # Backward untuk kelas tertentu
        self.model.zero_grad()
        output[0, class_idx].backward()

        # Hitung grad-cam
        pooled_gradients = torch.mean(self.gradients, dim=[0, 2, 3])
        activations = self.activations[0]

        for i in range(pooled_gradients.size(0)):
            activations[i, :, :] *= pooled_gradients[i]

        heatmap = torch.mean(activations, dim=0).cpu().numpy()
        heatmap = np.maximum(heatmap, 0)
        heatmap /= np.max(heatmap)

        return heatmap
```

```python
def visualize_all(image_path, mask_path, model, gradcam_layer, transform, alpha=0.5):
    # Load image dan mask
    original_bgr = cv2.imread(image_path)
    original_rgb = cv2.cvtColor(original_bgr, cv2.COLOR_BGR2RGB)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
    mask = cv2.resize(mask, (original_rgb.shape[1], original_rgb.shape[0]))

    # Normalize mask dan buat heatmap overlay
    mask_norm = cv2.normalize(mask, None, 0, 255, cv2.NORM_MINMAX)
    mask_heatmap = cv2.applyColorMap(mask_norm.astype(np.uint8), cv2.COLORMAP_JET)
    overlay_mask = cv2.addWeighted(original_bgr, 1 - alpha, mask_heatmap, alpha, 0)
    overlay_mask_rgb = cv2.cvtColor(overlay_mask, cv2.COLOR_BGR2RGB)

    # Grad-CAM
    gradcam = GradCAM(model, gradcam_layer)

    # Preprocessing input untuk model
    image_pil = Image.open(image_path).convert("RGB")
    input_tensor = transform(image_pil).unsqueeze(0).to(next(model.parameters()).device)

    cam = gradcam.generate(input_tensor)
    cam_resized = cv2.resize(cam, (original_rgb.shape[1], original_rgb.shape[0]))
    cam_heatmap = cv2.applyColorMap(np.uint8(255 * cam_resized), cv2.COLORMAP_JET)
    overlay_cam = cv2.addWeighted(original_bgr, 1 - alpha, cam_heatmap, alpha, 0)
    overlay_cam_rgb = cv2.cvtColor(overlay_cam, cv2.COLOR_BGR2RGB)

    # Plot semua
    plt.figure(figsize=(18, 10))
    plt.subplot(2, 3, 1)
    plt.imshow(original_rgb)
    plt.title("Gambar Asli")
    plt.axis('off')

    plt.subplot(2, 3, 2)
    plt.imshow(mask, cmap='gray')
    plt.title("Mask Segmentasi")
    plt.axis('off')

    plt.subplot(2, 3, 3)
    plt.imshow(overlay_mask_rgb)
    plt.title("Overlay Segmentasi")
    plt.axis('off')

    plt.subplot(2, 3, 4)
    plt.imshow(cam_resized, cmap='jet')
    plt.title("Heatmap Grad-CAM")
    plt.axis('off')

    plt.subplot(2, 3, 5)
    plt.imshow(overlay_cam_rgb)
    plt.title("Overlay Grad-CAM")
    plt.axis('off')

    plt.tight_layout()
    plt.savefig('heatmap_all_comparison.png', dpi=300)
    plt.show()
```

```python
def visualize_all_with_prediction(image_path, mask_path, model, gradcam_layer, transform, class_names=['Non-Cancer', 'Cancer'], alpha=0.5):
    import matplotlib.patches as patches

    # Load original image & mask
    original_bgr = cv2.imread(image_path)
    original_rgb = cv2.cvtColor(original_bgr, cv2.COLOR_BGR2RGB)
    mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
    mask = cv2.resize(mask, (original_rgb.shape[1], original_rgb.shape[0]))

    # Normalize mask dan buat heatmap
    mask_norm = cv2.normalize(mask, None, 0, 255, cv2.NORM_MINMAX)
    mask_heatmap = cv2.applyColorMap(mask_norm.astype(np.uint8), cv2.COLORMAP_JET)
    overlay_mask = cv2.addWeighted(original_bgr, 1 - alpha, mask_heatmap, alpha, 0)
    overlay_mask_rgb = cv2.cvtColor(overlay_mask, cv2.COLOR_BGR2RGB)

    # Grad-CAM
    gradcam = GradCAM(model, gradcam_layer)

    image_pil = Image.open(image_path).convert("RGB")
    input_tensor = transform(image_pil).unsqueeze(0).to(next(model.parameters()).device)

    with torch.no_grad():
        output = model(input_tensor)
        probabilities = torch.nn.functional.softmax(output, dim=1)
        confidence = torch.max(probabilities).item() * 100
        predicted_class = torch.argmax(probabilities, dim=1).item()
        predicted_label = class_names[predicted_class]

    cam = gradcam.generate(input_tensor)
    cam_resized = cv2.resize(cam, (original_rgb.shape[1], original_rgb.shape[0]))
    cam_heatmap = cv2.applyColorMap(np.uint8(255 * cam_resized), cv2.COLORMAP_JET)
    overlay_cam = cv2.addWeighted(original_bgr, 1 - alpha, cam_heatmap, alpha, 0)
    overlay_cam_rgb = cv2.cvtColor(overlay_cam, cv2.COLOR_BGR2RGB)

    # Plot
    plt.figure(figsize=(18, 10))

    plt.subplot(2, 3, 1)
    plt.imshow(original_rgb)
    plt.title("Gambar Asli")
    plt.axis('off')
    plt.text(5, 25, f"Pred: {predicted_label} ({confidence:.1f}%)", fontsize=12,
             bbox=dict(facecolor='yellow', alpha=0.7), color='black')

    plt.subplot(2, 3, 2)
    plt.imshow(mask, cmap='gray')
    plt.title("Mask Segmentasi")
    plt.axis('off')

    plt.subplot(2, 3, 3)
    plt.imshow(overlay_mask_rgb)
    plt.title("Overlay Segmentasi")
    plt.axis('off')

    plt.subplot(2, 3, 4)
    plt.imshow(cam_resized, cmap='jet')
    plt.title("Heatmap Grad-CAM")
    plt.axis('off')

    plt.subplot(2, 3, 5)
    plt.imshow(overlay_cam_rgb)
    plt.title("Overlay Grad-CAM")
    plt.axis('off')
    plt.text(5, 25, f"Pred: {predicted_label} ({confidence:.1f}%)", fontsize=12,
             bbox=dict(facecolor='yellow', alpha=0.7), color='black')

    plt.tight_layout()
    plt.savefig('heatmap_all_with_prediction.png', dpi=300)
    plt.show()

    return predicted_label, confidence
```

```python
image_path = "../../dataset/OralCancer_DatasetGabungan/CANCER/CANCER_020.jpeg"
mask_path = "../../dataset/OralCancer_DatasetGabungan/CANCER/CANCER_020.jpeg"
label, confidence = visualize_all_with_prediction(
    image_path=image_path,
    mask_path=mask_path,
    model=vgg_model,                    # model CNN kamu
    gradcam_layer=vgg_model.features[-1],  # layer target untuk Grad-CAM
    transform=transform_eval,              # transform untuk input
    class_names=['Non-Cancer', 'Cancer']
)
print(f"Predicted label: {label}, Confidence: {confidence:.2f}%")
```
